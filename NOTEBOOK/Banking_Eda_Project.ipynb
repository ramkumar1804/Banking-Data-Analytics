```python
bins = [0, 100000, 300000, float("inf")]
labels = ["Low", "Medium", "High"]

df["Income Band"] = pd.cut(
    df["Estimated Income"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df[["Estimated Income", "Income Band"]].head()
```

```python
df["Income Band"].value_counts().plot(kind="bar")

plt.title("Customer Distribution by Income Band")
plt.xlabel("Income Band")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
categorical_cols = [
    "BRId",
    "GenderId",
    "IAId",
    "Amount of Credit Cards",
    "Nationality",
    "Occupation",
    "Fee Structure",
    "Loyalty Classification",
    "Properties Owned",
    "Risk Weighting",
    "Income Band"
]

categorical_cols = [
    col for col in categorical_cols
    if col in df.columns
]

for col in categorical_cols:
    print(f"\nValue counts for '{col}':")
    display(df[col].value_counts(dropna=False))
```

```python
for col in categorical_cols:
    plt.figure(figsize=(9, 4))

    sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index
    )

    plt.title(f"Customer Distribution by {col}")
    plt.xlabel(col)
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
if "Joined Bank" in df.columns:
    df["Joined Bank"] = pd.to_datetime(
        df["Joined Bank"],
        format="%d-%m-%Y",
        errors="coerce"
    )

    print("Joined Bank data type:", df["Joined Bank"].dtype)
    display(df[["Joined Bank"]].head())
```

```python
numerical_cols = [
    "Age",
    "Estimated Income",
    "Superannuation Savings",
    "Credit Card Balance",
    "Bank Loans",
    "Bank Deposits",
    "Checking Accounts",
    "Saving Accounts",
    "Foreign Currency Account",
    "Business Lending"
]

numerical_cols = [
    col for col in numerical_cols
    if col in df.columns
    and pd.api.types.is_numeric_dtype(df[col])
]

df[numerical_cols].hist(
    figsize=(15, 10),
    bins=20
)

plt.suptitle("Distributions of Numerical Banking Variables")
plt.tight_layout()
plt.show()
```

```python
correlation_matrix = df[numerical_cols].corr()

display(correlation_matrix)
```

```python
plt.figure(figsize=(12, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Banking Variables")
plt.tight_layout()
plt.show()
```

```python
pairs_to_plot = [
    ("Bank Deposits", "Saving Accounts"),
    ("Checking Accounts", "Saving Accounts"),
    ("Checking Accounts", "Foreign Currency Account"),
    ("Age", "Superannuation Savings"),
    ("Estimated Income", "Checking Accounts"),
    ("Bank Loans", "Credit Card Balance"),
    ("Business Lending", "Bank Loans")
]

for x_col, y_col in pairs_to_plot:

    if x_col not in df.columns or y_col not in df.columns:
        continue

    if not (
        pd.api.types.is_numeric_dtype(df[x_col])
        and pd.api.types.is_numeric_dtype(df[y_col])
    ):
        continue

    plt.figure(figsize=(8, 6))

    sns.regplot(
        data=df,
        x=x_col,
        y=y_col,
        scatter_kws={"alpha": 0.4},
        line_kws={"color": "red"}
    )

    plt.title(f"{y_col} vs {x_col}")
    plt.tight_layout()
    plt.show()
```

```python
# Analyze categorical variables by nationality

categorical_cols = [
    "BRId",
    "GenderId",
    "IAId",
    "Amount of Credit Cards",
    "Nationality",
    "Occupation",
    "Fee Structure",
    "Loyalty Classification",
    "Properties Owned",
    "Risk Weighting"
]

# Keep only columns that exist in the dataset
categorical_cols = [
    col for col in categorical_cols
    if col in df.columns
]

# Plot categorical distributions by nationality
for predictor in categorical_cols:

    if predictor == "Nationality":
        continue

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        x=predictor,
        hue="Nationality"
    )

    plt.title(f"{predictor} Distribution by Nationality")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```
